# 🏏 IPL Data Analysis — 18 Seasons (2008–2025)
### Interactive Notebook Walkthrough
Explore top batsmen, bowlers, team stats, win %, venues, and IPL champions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
sns.set_theme(style='darkgrid')
%matplotlib inline
print('Libraries loaded ✅')
```

## 1️⃣ Load the Dataset

```python
matches = pd.read_csv('../data/matches.csv')
deliveries = pd.read_csv('../data/deliveries.csv')
print(f'Matches: {len(matches):,} | Deliveries: {len(deliveries):,}')
matches.head(3)
```

## 2️⃣ Overall Summary

```python
print('📊 OVERALL IPL STATS — 18 Seasons')
print('=' * 40)
print(f"Seasons       : {matches['season'].min()} – {matches['season'].max()}")
print(f"Total Matches : {len(matches):,}")
print(f"Total Balls   : {len(deliveries):,}")
print(f"Total Sixes   : {(deliveries['batsman_runs'] == 6).sum():,}")
print(f"Total Fours   : {(deliveries['batsman_runs'] == 4).sum():,}")
print(f"Total Wickets : {deliveries['is_wicket'].sum():,}")
print(f"Avg 1st Inn   : {matches['team1_score'].mean():.1f} runs")
```

## 3️⃣ Top 10 Run Scorers

```python
top_batsmen = (deliveries.groupby('batsman')['batsman_runs']
               .sum().reset_index()
               .rename(columns={'batsman_runs': 'Total_Runs'})
               .sort_values('Total_Runs', ascending=False).head(10))

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=top_batsmen, x='batsman', y='Total_Runs', palette='YlOrRd', ax=ax)
ax.set_xticklabels(ax.get_xticklabels(), rotation=30, ha='right')
ax.set_title('🏏 Top 10 Run Scorers — IPL All Time', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
top_batsmen
```

## 4️⃣ Top 10 Wicket Takers

```python
top_bowlers = (deliveries[deliveries['is_wicket'] == 1]
               .groupby('bowler')['is_wicket'].sum().reset_index()
               .rename(columns={'is_wicket': 'Wickets'})
               .sort_values('Wickets', ascending=False).head(10))

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=top_bowlers, x='bowler', y='Wickets', palette='Blues_r', ax=ax)
ax.set_xticklabels(ax.get_xticklabels(), rotation=30, ha='right')
ax.set_title('🎯 Top 10 Wicket Takers — IPL All Time', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
top_bowlers
```

## 5️⃣ Team Win Percentage

```python
played = pd.concat([matches['team1'], matches['team2']]).value_counts()
won = matches['winner'].value_counts()
win_pct = (won / played * 100).round(1).reset_index()
win_pct.columns = ['Team', 'Win%']
win_pct = win_pct.sort_values('Win%', ascending=False)

fig, ax = plt.subplots(figsize=(10, 6))
sns.barplot(data=win_pct, x='Win%', y='Team', palette='viridis', ax=ax)
ax.axvline(50, color='red', linestyle='--', alpha=0.7, label='50% threshold')
ax.set_title('🏆 Team Win Percentage — IPL 2008–2025', fontsize=14, fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()
win_pct
```

## 6️⃣ Season-wise Average Score Trends

```python
season_avg = matches.groupby('season')[['team1_score', 'team2_score']].mean().round(1)

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(season_avg.index, season_avg['team1_score'], marker='o', label='Batting First', linewidth=2)
ax.plot(season_avg.index, season_avg['team2_score'], marker='s', label='Chasing', linewidth=2)
ax.fill_between(season_avg.index, season_avg['team1_score'], season_avg['team2_score'], alpha=0.1)
ax.set_title('📈 Average Score Per Season', fontsize=14, fontweight='bold')
ax.set_xlabel('Season')
ax.set_ylabel('Average Runs')
ax.legend()
plt.tight_layout()
plt.show()
```

## 7️⃣ IPL Champions History

```python
champions = {
    2008:'Rajasthan Royals', 2009:'Kolkata Knight Riders', 2010:'Chennai Super Kings',
    2011:'Chennai Super Kings', 2012:'Kolkata Knight Riders', 2013:'Mumbai Indians',
    2014:'Kolkata Knight Riders', 2015:'Mumbai Indians', 2016:'Sunrisers Hyderabad',
    2017:'Mumbai Indians', 2018:'Chennai Super Kings', 2019:'Mumbai Indians',
    2020:'Mumbai Indians', 2021:'Chennai Super Kings', 2022:'Gujarat Titans',
    2023:'Chennai Super Kings', 2024:'Kolkata Knight Riders', 2025:'Royal Challengers Bangalore'
}
champ_df = pd.DataFrame(list(champions.items()), columns=['Year','Champion'])
title_count = champ_df['Champion'].value_counts().reset_index()
title_count.columns = ['Team','Titles']

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(data=title_count, x='Team', y='Titles', palette='magma', ax=ax)
ax.set_xticklabels(ax.get_xticklabels(), rotation=30, ha='right')
ax.set_title('👑 IPL Titles Won by Team', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
champ_df
```
